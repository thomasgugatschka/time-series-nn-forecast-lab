# Initial Data Analysis

This notebook is used for initial data analysis to check the contents of different datasets. Based on this review, data preprocessing and the data pipeline for the models will be set up.

```python
# Import necessary libraries
import os
import pandas as pd
from manager.path_manager import get_path
from IPython.display import display
```

```python
# Function to display data from a specific pickle file
def display_data_from_pickle(file_path):
    df = pd.read_pickle(file_path)
    print(f"Data from {os.path.basename(file_path)}:")
    with pd.option_context('display.max_columns', None):  # Display all columns
        display(df.head())  # Display the first few rows of the DataFrame
    print("\n")
```

```python
# Function to display the number of data points and the last 5 rows of the DataFrame
def display_data_summary(file_path):
    df = pd.read_pickle(file_path)
    num_data_points = df.shape[0]
    print(f"Data from {os.path.basename(file_path)}:")
    print(f"Number of data points: {num_data_points}")
    with pd.option_context('display.max_columns', None):  # Display all columns
        display(df.tail(5))  # Display the last 5 rows of the DataFrame
    if 'Date/Time (LST)' in df.columns:
        df['Time (h)'] = pd.to_datetime(df['Date/Time (LST)']).dt.hour
        print("Time in hours:")
        display(df[['Date/Time (LST)', 'Time (h)']].tail(5))
    print("\n")
```

```python
# Specify the directory containing the pickle files
pickle_directory = get_path('data_dir', 'pickles')
files = [f for f in os.listdir(pickle_directory) if f.endswith('.pkl')]
```

```python
# Display available files
print("Available pickle files:")
for i, file in enumerate(files):
    print(f"{i}: {file}")
```

```python
# Select a file to display
file_index = int(input("Enter the index of the file to display: "))
selected_file = os.path.join(pickle_directory, files[file_index])
```

```python
# Display data from the selected file
display_data_from_pickle(selected_file)
```

```python
# Display data summary from the selected file
display_data_summary(selected_file)
```
